### Da dove vengono questi dati e cosa non possiamo farci
Dodici film **inventati**, descritti da due numeri: minuti di scene d'azione e numero di battute comiche. Il genere è stato deciso da noi. Servono solo a capire come ragiona kNN.

```python
import math
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler

# [minuti di scene d'azione, numero di battute comiche]
film = [[80, 5], [75, 10], [90, 2], [70, 8], [85, 12],
        [10, 60], [5, 70], [15, 55], [20, 65], [8, 50],
        [45, 30], [50, 35]]
genere = ["azione"] * 5 + ["commedia"] * 5 + ["azione", "commedia"]
```

**Prevedi.** Qual è la distanza tra il film `[80, 5]` e il film `[75, 10]`? Calcolala a mano prima di eseguire.

*La mia previsione:* …

```python
def distanza(a, b):
    somma = 0
    for i in range(len(a)):
        somma = somma + (a[i] - b[i]) ** 2
    return math.sqrt(somma)

round(distanza([80, 5], [75, 10]), 2)
```

```python
def classifica(nuovo, esempi, etichette, k):
    coppie = []
    for i in range(len(esempi)):
        coppie.append((distanza(nuovo, esempi[i]), etichette[i]))
    coppie.sort()                      # dal più vicino al più lontano
    voti = {}
    for d, e in coppie[:k]:            # solo i k più vicini votano
        voti[e] = voti.get(e, 0) + 1
    migliore = None
    for e in voti:
        if migliore is None or voti[e] > voti[migliore]:
            migliore = e
    return migliore
```

**Prevedi.** Con k = 3, di che genere saranno i film `[60, 20]`, `[30, 40]`, `[48, 33]`, `[47, 32]`?

*La mia previsione:* …

```python
nuovi = [[60, 20], [30, 40], [48, 33], [47, 32]]
print("da zero:     ", [classifica(n, film, genere, 3) for n in nuovi])
libreria = KNeighborsClassifier(n_neighbors=3).fit(film, genere)
print("scikit-learn:", [str(p) for p in libreria.predict(nuovi)])
```

**Indaga.** Le due righe coincidono? Quante righe di codice abbiamo scritto noi, e quante ne servono con scikit-learn?

**Prevedi.** E con k = 2? Guarda i due vicini più vicini del film `[48, 33]`: che cosa succede?

*La mia previsione:* …

```python
print(sorted((round(distanza([48, 33], f), 2), g) for f, g in zip(film, genere))[:2])
print("da zero:     ", classifica([48, 33], film, genere, 2))
print("scikit-learn:", KNeighborsClassifier(n_neighbors=2).fit(film, genere).predict([[48, 33]])[0])
```

**Indaga.** Perché i due risultati sono diversi? Chi ha "ragione"? (Suggerimento: rileggi l'ultimo ciclo di `classifica`.)

**Prevedi.** Se misurassimo l'azione in **secondi** invece che in minuti, il film `[30, 10]` cambierebbe genere?

*La mia previsione:* …

```python
film_secondi = [[a * 60, b] for a, b in film]
print("in minuti: ", classifica([30, 10], film, genere, 3))
print("in secondi:", classifica([30 * 60, 10], film_secondi, genere, 3))
```

```python
# Con StandardScaler ogni colonna ha media 0 e deviazione standard 1: l'unità di misura non conta più
for dati, nuovo, nome in [(film, [30, 10], "minuti"), (film_secondi, [30 * 60, 10], "secondi")]:
    scaler = StandardScaler().fit(dati)
    modello = KNeighborsClassifier(n_neighbors=3).fit(scaler.transform(dati), genere)
    print(nome, "con scaling:", modello.predict(scaler.transform([nuovo]))[0])
```

### Modifica
Cambia `classifica` in modo che, in caso di parità, vinca il genere del vicino più vicino **in modo esplicito** (non per caso). Poi prova k = 4.
