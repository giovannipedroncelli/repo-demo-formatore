```python
import numpy as np
import pandas as pd
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.dummy import DummyClassifier
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, recall_score

# Dati di esempio (inventati): 1000 clienti, 30% che abbandona il servizio (classe 1)
X, y = make_classification(n_samples=1000, n_features=10, n_informative=3,
                           weights=[0.7, 0.3], flip_y=0.1, random_state=1)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=1, stratify=y)
```

## Caso 1 — "Il modello è perfetto"

```python
albero = DecisionTreeClassifier(random_state=1).fit(X_train, y_train)
print("Accuratezza del modello:", accuracy_score(y_train, albero.predict(X_train)))
```

## Caso 2 — "Abbiamo trovato le variabili giuste"

```python
tabella = pd.DataFrame(X, columns=[f"v{i}" for i in range(10)])
tabella["abbandona"] = y
caratteristiche = tabella.copy()          # tutte le colonne come feature
a, b, c, d = train_test_split(caratteristiche, y, test_size=0.25, random_state=1)
print("Accuratezza:", accuracy_score(d, DecisionTreeClassifier(random_state=1).fit(a, c).predict(b)))
```

## Caso 3 — "0,78 è un ottimo risultato"

```python
albero = DecisionTreeClassifier(random_state=1).fit(X_train, y_train)
print("Accuratezza sul test:", round(accuracy_score(y_test, albero.predict(X_test)), 3))
print("Ottimo risultato: il modello funziona.")
```

## Caso 4 — "Il modello più semplice va già benissimo"

```python
semplice = DummyClassifier(strategy="most_frequent").fit(X_train, y_train)
print("Accuratezza:", accuracy_score(y_test, semplice.predict(X_test)))
print("Con due righe di codice abbiamo quasi il 70%: lo usiamo per individuare chi abbandona.")
```

## Caso 6 — "Quattro segmenti di clientela"

```python
rng = np.random.default_rng(0)
reddito = np.concatenate([rng.normal(20000, 3000, 50), rng.normal(60000, 5000, 50)])
eta = np.concatenate([rng.normal(25, 3, 25), rng.normal(55, 4, 25), rng.normal(25, 3, 25), rng.normal(55, 4, 25)])
clienti = pd.DataFrame({"reddito": reddito, "eta": eta})
clienti["segmento"] = KMeans(n_clusters=4, n_init=10, random_state=0).fit_predict(clienti[["reddito", "eta"]])
clienti.groupby("segmento")[["reddito", "eta"]].mean().round(0)
```
