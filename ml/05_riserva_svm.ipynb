### Da dove vengono questi dati e cosa non possiamo farci
Punti **generati** da scikit-learn (`make_blobs`, `make_moons`) con un seme fisso: non rappresentano nulla di reale. Servono a vedere la forma del confine che la SVM impara.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs, make_moons
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

def disegna(modello, X, y, titolo, asse):
    xx, yy = np.meshgrid(np.linspace(X[:, 0].min() - 1, X[:, 0].max() + 1, 200),
                         np.linspace(X[:, 1].min() - 1, X[:, 1].max() + 1, 200))
    z = modello.decision_function(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
    asse.contourf(xx, yy, z > 0, alpha=0.2)
    asse.contour(xx, yy, z, levels=[-1, 0, 1], linestyles=["--", "-", "--"], colors="k")
    asse.scatter(X[:, 0], X[:, 1], c=y, s=20)
    asse.scatter(*modello.support_vectors_.T, s=80, facecolors="none", edgecolors="k")
    asse.set_title(titolo)
```

## 1. Due gruppi separabili: l'iperpiano

```python
X, y = make_blobs(n_samples=100, centers=2, random_state=6)
lineare = SVC(kernel="linear", C=1).fit(X, y)
print("vettori di supporto:", len(lineare.support_vectors_))
fig, asse = plt.subplots(figsize=(5, 4))
disegna(lineare, X, y, "kernel lineare, C = 1", asse)
plt.show()
```

**Indaga.** La linea continua è il confine (l'iperpiano, in due dimensioni una retta); le tratteggiate delimitano il **margine**. I punti cerchiati sono i vettori di supporto: perché sono proprio quelli?

**Prevedi.** Con C = 0,01 (margine «morbido») i vettori di supporto saranno di più o di meno? E con C = 100?

*La mia previsione:* …

```python
fig, assi = plt.subplots(1, 2, figsize=(10, 4))
for asse, c in zip(assi, [0.01, 100]):
    m = SVC(kernel="linear", C=c).fit(X, y)
    print("C =", c, "→ vettori di supporto:", len(m.support_vectors_))
    disegna(m, X, y, f"C = {c}", asse)
plt.show()
```

## 2. Due gruppi che una retta non separa: il kernel

```python
X2, y2 = make_moons(n_samples=200, noise=0.2, random_state=0)
X_train, X_test, y_train, y_test = train_test_split(X2, y2, test_size=0.25, random_state=0, stratify=y2)
```

**Prevedi.** Sulle due «mezzelune», quale accuratezza avrà il kernel lineare? E il kernel `rbf`?

*La mia previsione:* …

```python
fig, assi = plt.subplots(1, 2, figsize=(10, 4))
for asse, kernel in zip(assi, ["linear", "rbf"]):
    m = SVC(kernel=kernel).fit(X_train, y_train)
    print(kernel, "→ accuratezza sul test:", round(accuracy_score(y_test, m.predict(X_test)), 3))
    disegna(m, X2, y2, "kernel " + kernel, asse)
plt.show()
```

**Indaga.** Il codice è cambiato di una parola (`kernel`). Che cosa è cambiato nel confine?
